# Automated Academic Research Assistant Multi-Agent System
### **Pipeline with RAG, Fact-Check Critic, and Contradiction Detection**

**Objective:**
To design, build, and demonstrate a multi-agent system that automates an enterprise academic literature review workflow by applying:
1. **Defined Coordination Architecture:** Pipeline with RAG and a Critic Architecture.
2. **Framework-Based Agent Development:** Microsoft Semantic Kernel.
3. **Grounded Retrieval & Shared Memory:** Corpus RAG with verifiable claim entailment ledger.
4. **Collaborative Task Execution:** 6 specialized modular agents with >= 2 real plain Python tools per agent.
5. **Cross-Source Contradiction Analysis:** Contradiction Agent detecting and presenting dual divergent perspectives.

````python
# Step 1: Import Environment and Semantic Kernel Multi-Agent Modules
import os
import sys
import json
if os.path.isdir('AgenticCA2-main') and 'AgenticCA2-main' not in sys.path:
    sys.path.insert(0, os.path.abspath('AgenticCA2-main'))
from dotenv import load_dotenv
load_dotenv()

import semantic_kernel as sk
print(f"Semantic Kernel Version: {sk.__version__}")

from app.agents import (
    AcademicPipelineOrchestrator,
    QueryAgent,
    RetrievalAgent,
    SummaryAgent,
    FactCheckAgent,
    ContradictionAgent,
    WriterAgent,
    DEFAULT_PAPER_CORPUS
)
from app.agents.tools import (
    QueryAgentTools,
    RetrievalAgentTools,
    SummaryAgentTools,
    FactCheckAgentTools,
    ContradictionAgentTools,
    WriterAgentTools
)
print("All multi-agent modules and tools successfully loaded.")
````

````python
# Step 2: Verify Tool Roster Contract (>= 2 Real Plain Python Tools Per Working Agent)
agents_tools = {
    "Query Agent": [m for m in dir(QueryAgentTools) if not m.startswith("_")],
    "Retrieval Agent": [m for m in dir(RetrievalAgentTools) if not m.startswith("_")],
    "Summary Agent": [m for m in dir(SummaryAgentTools) if not m.startswith("_")],
    "Fact-Check Agent (Critic)": [m for m in dir(FactCheckAgentTools) if not m.startswith("_")],
    "Contradiction Agent": [m for m in dir(ContradictionAgentTools) if not m.startswith("_")],
    "Writer Agent": [m for m in dir(WriterAgentTools) if not m.startswith("_")]
}

print("================ AGENT TOOL VERIFICATION ROSTER ================")
for agent, tools in agents_tools.items():
    print(f"\n[Agent] {agent} (Total Tools: {len(tools)}):")
    for t in tools:
        print(f"   * Tool: {t}()")
    assert len(tools) >= 2, f"{agent} has less than 2 tools!"
print("\n[PASSED] All agents strictly satisfy the plain Python tool contract.")
````

````python
# Step 3: SCENARIO 1 - Standard Grounded Research Pipeline Run
orchestrator = AcademicPipelineOrchestrator()

inquiry = "What are the risks and coordination patterns in autonomous multi-agent workflows?"
print(f"Research Question: {inquiry}\n")

result_scenario_1 = orchestrator.run_pipeline(question=inquiry)

print(f"Run ID: {result_scenario_1['run_id']}")
print(f"Pipeline Architecture: {result_scenario_1['architecture']}")
print(f"Execution Duration: {result_scenario_1['metrics']['total_duration_ms']} ms\n")

print("================ DETAILED MULTI-AGENT TRACE ================")
for idx, step in enumerate(result_scenario_1["agent_trace"], 1):
    print(f"Step {idx:02d} | Agent: {step['agent']:<26} | Tool: {step['tool']}() [{step['duration_ms']} ms]")
    print(f"        Output snippet: {str(step['output'])[:120]}...")
````

````python
# Step 4: Display Grounded Literature Summary & Per-Claim Citations
report = result_scenario_1["final_report"]
print("================ FINAL SOURCED LITERATURE REPORT ================")
print(report["markdown_report"])

print("\n================ AUDITED CLAIM VERIFICATION LEDGER ================")
for claim in result_scenario_1["retained_claims"]:
    print(f"{claim['claim_id']}: {claim['statement']}")
    print(f"   --> Citation: {claim['cited_tag']} | Source: {claim['source_paper']}")
    print(f"   --> Critic Status: {claim['status']} ({int(claim['confidence']*100)}% confidence)\n")
````

````python
# Step 5: SCENARIO 2 - Critic Failure Handling & Hallucination Pruning
print("Executing Scenario 2: Injecting an unsupported synthetic claim into the pipeline...")
result_scenario_2 = orchestrator.run_pipeline(
    question="What are the empirical trade-offs of deep residual learning?",
    inject_unsupported_claim=True
)

print(f"Total Candidate Claims: {result_scenario_2['metrics']['candidate_claims']}")
print(f"Retained Supported Claims: {result_scenario_2['metrics']['retained_claims']}")
print(f"Dropped Unsupported Claims: {result_scenario_2['metrics']['dropped_claims']}\n")

print("================ CRITIC DROPPED CLAIMS AUDIT LEDGER ================")
for dc in result_scenario_2["dropped_claims"]:
    print(f"[DROPPED] Claim ID: {dc['claim_id']}")
    print(f"  Statement: {dc['statement']}")
    print(f"  Reason: {dc['critic_verdict']}\n")
````

````python
# Step 6: SCENARIO 3 - Contradiction Agent (Cross-Source Divergence)
debate_query = "Compare centralized orchestration vs decentralized blackboard architectures for autonomous agent systems."
print(f"Executing Scenario 3: {debate_query}\n")

result_scenario_3 = orchestrator.run_pipeline(question=debate_query)
contradictions = result_scenario_3["contradictions"]

print(f"Has Contradictions: {contradictions['has_contradictions']}")
print(f"Contradiction Count: {contradictions.get('count', 0)}\n")

for d in contradictions.get("dossiers", []):
    print(f"--- Debate Topic: {d['topic']} ---")
    print(f"* Position A ({d['position_a']['source']}):")
    print(f"    {d['position_a']['argument']}")
    print(f"* Position B ({d['position_b']['source']}):")
    print(f"    {d['position_b']['argument']}")
    print(f"* Architectural Significance: {d['significance']}\n")
````

````python
# Step 7: Inspect Append-Only Run Log (Audit Trail)
log_path = getattr(orchestrator, 'log_file_path', None) or (os.path.abspath('run_log.jsonl') if os.path.exists('run_log.jsonl') else os.path.abspath(os.path.join('AgenticCA2-main', 'run_log.jsonl')))
print(f"Reading Append-Only Run Log from: {log_path}\n")

with open(log_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

print(f"Total Runs Logged: {len(lines)}\n")
for l in lines[-4:]:
    entry = json.loads(l.strip())
    print(f"Run: {entry['run_id']} | Time: {entry['timestamp'][:19]} | Kept: {entry['retained_claims_count']} | Dropped: {entry['dropped_claims_count']} | Latency: {entry['duration_ms']} ms")
````

## Architecture Diagram (Mermaid)

```mermaid
graph TD
    User([Researcher Inquiry]) --> A[Query Agent]
    A -->|Sub-Queries & Lexicon| B[Retrieval Agent]
    B -->|Grounded Corpus RAG [P1, P2...]| C[Summary Agent]
    C -->|Themed Synthesis & Candidate Claims| D{Fact-Check Agent / Critic}
    D -->|Unsupported Claims| X[Pruned / Dropped Ledger]
    D -->|Supported Claims| E[Contradiction Agent]
    B -->|Passage Pairs| E
    E -->|Dual-Perspective Dossier| F[Writer Agent]
    D -->|Verified Claims| F
    F --> G[Sourced Literature Summary]
    F --> H[Append-Only Run Log - run_log.jsonl]
```

### Why this Architecture Fits:
- **Pipeline with RAG and Critic Architecture** prevents context drift and hallucination cascades by imposing an independent verification gate before publication.
- The **Fact-Check Critic** enforces strict NLI token entailment against memory passages.
- The **Contradiction Agent** ensures academic objectivity by presenting the dual positions on contested empirical questions.
